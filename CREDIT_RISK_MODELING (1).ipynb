### regex=False tells pandas not to treat the pattern as a regular expression, and instead treat it as a literal text match.

## 🔍 Why this matters

str.replace() by default uses regex = True, meaning the pattern is treated as a regular expression.

This can cause unexpected behavior.

### ✅ **`regex=False` means: "replace the text exactly as it is."**

Pandas will NOT treat the text as a special pattern.

### ❌ Without `regex=False`

Symbols like `+`, `?`, `.`, `*` are treated as **special regex characters**, causing errors or wrong replacements.

---

### ⭐ **Example**

```python
s.str.replace("+", "")
```

❌ Fails because `+` is a special regex symbol.

```python
s.str.replace("+", "", regex=False)
```

✔ Works because it treats `+` as normal text.

---

So:

### 👉 **Use `regex=False` when your replace text contains special characters like +, *, ?, ., ( )**

````python
# List of numeric columns to visualize
cols = ['loan_amnt', 'funded_amnt', 'funded_amnt_inv', 'int_rate', 'installment']

# Creating a figure with 3 rows and 2 columns of subplots
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(16, 12))

# Flattening 2D axes array to 1D for easier indexing
axes = axes.flatten()

# Loop through each column using index (i) and column name (col)
for i, col in enumerate(cols):

    # Convert column values to numeric (removes errors), drop non-numeric values (NaN)
    data = pd.to_numeric(df3[col], errors='coerce').dropna()

    # Select the subplot axis for histogram
    ax1 = axes[i]

    # -----------------------------
    # PLOT 1: HISTOGRAM (BAR GRAPH)
    # -----------------------------
    sns.histplot(
        data=data,            # numeric dataset
        bins=40,              # number of bars
        color="skyblue",      # bar color
        edgecolor="black",    # border of bars
        alpha=0.6,            # transparency
        ax=ax1                # plot on primary axis
    )

    # Setting labels and title for histogram
    ax1.set_xlabel(col)                      # x-axis label
    ax1.set_ylabel("Frequency")              # y-axis label
    ax1.set_title(f"Histogram + KDE of {col}")  # plot title

    # ----------------------------------------
    # CREATE SECONDARY AXIS FOR KDE ON SAME PLOT
    # ----------------------------------------
    ax2 = ax1.twinx()   # This shares the x-axis but has separate y-axis for density

    # -----------------------------
    # PLOT 2: KDE CURVE (SMOOTH LINE)
    # -----------------------------
    sns.kdeplot(
        data=data,       # numeric dataset
        color="red",     # KDE line color
        linewidth=2,     # thickness of KDE line
        bw_adjust=0.3,   # smoothing factor (lower = more wiggly)
        ax=ax2           # plot KDE on secondary axis
    )

    # Y-axis label for KDE curve
    ax2.set_ylabel("Density")

    # Remove background grid for KDE axis only
    ax2.grid(False)

# -------------------------------
# Remove the last subplot (empty)
# -------------------------------
fig.delaxes(axes[-1])   # deletes unused 6th subplot space

# Adjust spacing between subplots
plt.tight_layout()

# Show the final plot
plt.show()
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_kde_box(df, column):
    """
    Plots a KDE plot and a Boxplot for a given numeric column in one row.

    Parameters:
        df (DataFrame): Input dataframe
        column (str): Column to plot
    """

    plt.figure(figsize=(12, 5))  # width, height

    # --- KDE Plot ---
    plt.subplot(1, 2, 1)
    sns.kdeplot(data=df, x=column)
    plt.title(f'KDE Plot of {column}')
    plt.xlabel(column)

    # --- Box Plot ---
    plt.subplot(1, 2, 2)
    sns.boxplot(data=df, y=column)
    plt.title(f'Boxplot of {column}')
    plt.ylabel(column)

    plt.tight_layout()
    plt.show()
````

````python

def cap_outliers(df, column):
    
    # Calculate Q1, Q2 (median), Q3
    q1, q2, q3,q96,q97,q98,q99 = df[column].quantile([0.25, 0.50, 0.75,0.96,0.97,0.98,0.99])

    # Calculate IQR
    IQR = q3 - q1

    # Calculate limits
    upperlimit = q3 + 1.5 * IQR
    lowerlimit = q1 - 1.5 * IQR

    # Cap upper outliers
    df[column] = np.where(df[column] > upperlimit, upperlimit, df[column])

    # Cap lower outliers
    df[column] = np.where(df[column] < lowerlimit, lowerlimit, df[column])

    # Fill missing values with median
    df[column] = df[column].fillna(q2)

    return q1, q2, q3, q96, q97, q98, q99, IQR

````

````python
def important_cat_variables(
    df_raw,
    df_encoded,
    variable,
    target_variable,
    threshold=2
):
    """
    Calculate IV for a categorical variable and
    drop encoded columns if IV <= threshold
    """

    # ---------- Safety checks ----------
    if variable not in df_raw.columns:
        raise ValueError(f"{variable} not found in raw dataframe")

    if target_variable not in df_raw.columns:
        raise ValueError(f"{target_variable} not found in raw dataframe")

    # ---------- IV calculation ----------
    df4 = df_raw[[variable, target_variable]].copy()

    summary = (
        df4
        .groupby(variable)[target_variable]
        .agg(count='count', good_rate='mean')
        .reset_index()
    )

    summary['bad_rate'] = 1 - summary['good_rate']
    summary['good'] = (summary['count'] * summary['good_rate']).round()
    summary['bad'] = summary['count'] - summary['good']

    summary['prop_good'] = (summary['good'] / summary['good'].sum())+0.5
    summary['prop_bad'] = (summary['bad'] / summary['bad'].sum())+0.5

    summary['WoE'] = np.log(
        np.where(summary['prop_bad'] == 0, 0,
                 summary['prop_good'] / summary['prop_bad'])
    )

    summary['IV'] = (summary['prop_good'] - summary['prop_bad']) * summary['WoE'] * 100

    summary['total_iv'] = summary['IV'].sum()

    # ---------- Decision ----------
    drop_columns = []

    if summary['total_iv'].iloc[0] <= threshold:
        drop_columns = [
            col for col in df_encoded.columns
            if col.startswith(variable)
        ]
        # df_encoded = df_encoded.drop(columns=drop_columns)

    return summary, summary['total_iv'].iloc[0], drop_columns
````

````python
def plot_by_woe(df_woe, rotation_of_x_axis_labels=0, show_point_values=True):
    # x labels (your first column)
    x = np.array(df_woe.iloc[:, 0].astype(str))
    # y values (WoE column)
    y = df_woe["WoE"].astype(float).to_numpy()

    fig, ax = plt.subplots(figsize=(18, 6))
    ax.plot(x, y, marker="o", linestyle="--", color="k")

    ax.set_xlabel(df_woe.columns[0])
    ax.set_ylabel("Weight of Evidence")
    ax.set_title(f"Weight of Evidence by {df_woe.columns[0]}")
    ax.tick_params(axis="x", rotation=rotation_of_x_axis_labels)

    # --- add value labels on each point (2 decimals) ---
    if show_point_values:
        for xi, yi in zip(x, y):
            ax.annotate(
                f"{yi:.2f}",            # 2 decimal digits [web:22]
                xy=(xi, yi),
                xytext=(0, 8),          # shift label up a bit
                textcoords="offset points",
                ha="center",
                va="bottom",
                fontsize=9,
                color="black"
            )  # annotate usage [web:23]

    plt.tight_layout()
    plt.show()

# usage:
# plot_by_woe(iv_table, rotation_of_x_axis_labels=0)
````
